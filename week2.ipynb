```python
import random
from datetime import datetime, timedelta
DEVICE_TYPES = ['Temperature', 'Humidity', 'Pressure', 'CO2']
LOCATIONS = ['Server_Room_A', 'Warehouse_B', 'Lab_C', 'Office_Floor_1']

READING_RANGES = {
    'Temperature': (15.0, 35.0),
    'Humidity': (30.0, 70.0),
    'Pressure': (950.0, 1050.0),
    'CO2': (400.0, 1200.0)
}
base_time = datetime(2026, 7, 20, 12, 0, 0)

raw_sensor_data = []

for i in range(1, 101)
    device_id = f"DEV{i:03d}"
    device_type = random.choice(DEVICE_TYPES)
    location = random.choice(LOCATIONS)
    timestamp = base_time + timedelta(minutes=i * 5)
    timestamp_str = timestamp.strftime('%Y-%m-%d %H:%M:%S')
    min_val, max_val = READING_RANGES[device_type]
    value = round(random.uniform(min_val, max_val), 2
    raw_sensor_data.append((timestamp_str, device_id, device_type, location, value
for reading in raw_sensor_data[:5]:
    print(reading)
```

```python
temperature_readings = []

for reading in raw_sensor_data:
    timestamp_str, device_id, device_type, location, value = reading
    if device_type in ['Temperature', 'Temperature_Sensor']
        reading_dict = {
            'timestamp': timestamp_str,
            'device_id': device_id,
            'location': location,
            'temperature_celsius': value
        }
        temperature_readings.append(reading_dict)
for t_reading in temperature_readings[:3]:
    print(t_reading)
```

```python
device_stats = {}

for reading in raw_sensor_data:
    # Unpack the tuple
    _, _, device_type, _, value = reading

    # If we haven't seen this device type yet, initialize it with the current value
    if device_type not in device_stats:
        device_stats[device_type] = (value, value)
    else:
        # Get the current min and max
        current_min, current_max = device_stats[device_type]

        # Update them if the new value breaks the record
        new_min = min(current_min, value)
        new_max = max(current_max, value)

        device_stats[device_type] = (new_min, new_max)

# Print the final stats
for dev_type, stats in device_stats.items():
    print(f"{dev_type}: Min = {stats[0]}, Max = {stats[1]}")
```

```python
def get_readings_by_location(readings_list, target_location):
    """
    Filters a list of sensor readings and returns only those
    matching the specified target location.
    """
    # reading[3] corresponds to the 'location' element in the tuple
    return [reading for reading in readings_list if reading[3] == target_location]


# --- Quick Test ---
# Let's try finding all readings for 'Server_Room_A'
server_room_readings = get_readings_by_location(raw_sensor_data, 'Server_Room_A')

print(f"Found {len(server_room_readings)} readings in Server_Room_A.")
if server_room_readings:
    print(f"Sample reading: {server_room_readings[0]}")
```

```python
def calculate_average_temperature(temp_data_list):
    """
    Calculates the average temperature from a list of temperature dictionaries.
    Returns 0.0 if the list is empty.
    """
    # Guard clause to handle an empty list safely
    if not temp_data_list:
        return 0.0

    # Sum up all the temperature values using a generator expression
    total_temp = sum(item['temperature_celsius'] for item in temp_data_list)

    # Calculate the average and round to 2 decimal places
    average_temp = total_temp / len(temp_data_list)
    return round(average_temp, 2)


# --- Quick Test ---
# Testing with the 'temperature_readings' list built in Task 1.2
avg_temp = calculate_average_temperature(temperature_readings)
print(f"Average Temperature: {avg_temp}°C")

# Testing with an empty list to verify the fallback condition
print(f"Empty list fallback: {calculate_average_temperature([])}")
```

```python
def generate_report_line(reading_dict):
    """
    Takes a single temperature reading dictionary and returns a
    formatted report string.
    """
    return (
        f"[{reading_dict['timestamp']}] "
        f"Device {reading_dict['device_id']} at {reading_dict['location']}: "
        f"{reading_dict['temperature_celsius']}°C"
    )


# --- Quick Test ---
# Test using a mock dictionary matching the Task 1.2 structure
sample_dict = {
    'timestamp': '2026-07-20 12:05:00',
    'device_id': 'DEV001',
    'location': 'Server_Room_A',
    'temperature_celsius': 23.45
}

print(generate_report_line(sample_dict))
```
